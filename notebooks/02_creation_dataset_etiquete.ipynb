 Étape 3 : Étiquetage Automatique

Amélioration du Lexique

Ajoutez 3 mots positifs et 3 mots négatifs au lexique.

```python
# 03_etiquetage_auto.py - Partie 1 (Méthode par règles)

# EXPLICATION : Classe basée sur des Dictionnaires
# Cette approche utilise des lexiques de mots positifs/négatifs avec des poids associés.
# C'est une méthode transparente et rapide, mais rigide.
class EtiqueteurRegles:
    def __init__(self):
        # Lexique enrichi avec la solution de l'Exercice 2 et l'ajout de 'bon'
        self.mots_positifs = {
            'bon': 1, 'délicieux': 2, 'impeccable': 2, 'excellent': 3,
            'génial': 2, 'super': 2, 'parfait': 3,
            'fantastique': 3, 'succulent': 2, 'aimable': 1
        }
        self.mots_negatifs = {
            'horrible': 3, 'cher': 1, 'désagréable': 2,
            'interminable': 2, 'catastrophique': 3,
            'mauvais': 2, 'nul': 2, 'lent': 1, 'arnaqué': 3, 'décevant': 2
        }

    def analyser(self, texte):
        score = 0
        texte_lower = texte.lower()

        # EXPLICATION : Calcul du Score
        # Le score est calculé par la somme des poids des mots trouvés.
        for mot, poids in self.mots_positifs.items():
            if mot in texte_lower: score += poids
        for mot, poids in self.mots_negatifs.items():
            if mot in texte_lower: score -= poids

        # EXPLICATION : Décision de Classification
        # Les seuils (ici > 2 et <= -2) sont définis pour séparer les classes.
        if score > 2: return "positif"
        elif score <= -2: return "négatif"
        else: return "neutre"
```

Méthode 2 : Modèle Naive Bayes

```python
# 03_etiquetage_auto.py - Partie 2 (Modèle simple)

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
import numpy as np
# EXPLICATION : Naive Bayes
# Cette méthode apprend des motifs à partir de données d'entraînement (même minimales).

# Données d'entraînement minimales
train_texts  = ["j'adore ce plat", "service parfait", "super expérience", "je déteste l'attente", "horrible", "c'est nul"]
train_labels = ['positif', 'positif', 'positif', 'négatif', 'négatif', 'négatif']

# 1. Vectorisation : Convertit le texte en vecteurs numériques (fréquence des mots)
vectorizer = CountVectorizer()
X_train = vectorizer.fit_transform(train_texts)

# 2. Entraînement du modèle
model = MultinomialNB()
model.fit(X_train, train_labels)

def predire_sentiment_nb(texte):
    # 3. Transformation du nouveau texte avec le vocabulaire appris
    X_test = vectorizer.transform([texte])

    # 4. Prédiction et Probabilités
    prediction = model.predict(X_test)[0]
    # predict_proba retourne la probabilité de chaque classe (utile pour la confiance)
    probabilites = model.predict_proba(X_test)[0]

    # Trouve la confiance du label prédit
    max_proba = np.max(probabilites)

    return prediction, max_proba

# NOTE : Pour l'usage dans l'Étape 4, on utiliserait le modèle NB pour annoter
# un grand volume de données non étiquetées, en récupérant le label et la confiance.
```

Étape 4 : Création du Dataset Final

Objectif : Fusionner les annotations manuelles (haute qualité) avec les annotations automatiques (grande quantité) pour créer un dataset final.



Stratégie de Fusion :

La stratégie de fusion est basée sur la priorité aux données humaines, complétées par des prédictions automatiques si la confiance est élevée (supérieure à 0.7)

```python
# 04_dataset_final.py - Stratégie de Fusion

# EXPLICATION : Logique de Fusion
# La fusion intelligente permet de créer un dataset plus grand tout en contrôlant la qualité.
# 1. Humain toujours prioritaire (qualité 'haute').
# 2. Auto est ajouté seulement si la 'confiance' est > 0.7 (qualité 'moyenne').
def fusionner_annotations(manuelles, auto):
    dataset_final = []

    # Priorité 1 : Annotations manuelles
    for ann in manuelles:
        ann['source'] = 'manuelle'
        ann['qualite'] = 'haute'
        dataset_final.append(ann)

    # Priorité 2 : Annotations automatiques à haute confiance
    for ann in auto:
        # EXPLICATION : Seuil de Confiance
        # On n'accepte la prédiction que si le modèle est suffisamment sûr.
        if ann.get('confiance', 0) > 0.7:
            ann['source'] = 'automatique'
            ann['qualite'] = 'moyenne'
            # Note : ann['id'] doit être mis à jour pour éviter les doublons avec 'manuelles'
            dataset_final.append(ann)

    # Logique de fusion :
    # 1. Humain toujours prioritaire
    # 2. Auto haute confiance complète
    # 3. Auto basse confiance à vérifier
    # 4. Conflits revue humaine

    return dataset_final
```

Division du Dataset (Train/Test/Validation)

```python
# 04_dataset_final.py - Division
import pandas as pd
from sklearn.model_selection import train_test_split
import json

# --- Chargement du DataFrame df (Solution pour l'erreur 'df' non défini) ---
# Dans un scénario complet, 'df' serait le résultat de la fusion des annotations.
# Pour rendre ce bloc de code exécutable, nous chargeons les annotations manuelles.
try:
    with open('data/avis_etiquetes.json', 'r', encoding='utf-8') as f:
        annotations = json.load(f)
    df = pd.DataFrame(annotations)
    print("DataFrame 'df' chargé depuis 'avis_etiquetes.json'.")
except FileNotFoundError:
    print("Erreur: 'data/avis_etiquetes.json' non trouvé. Veuillez exécuter les étapes précédentes.")
    # Créer un DataFrame vide ou de simulation si le fichier n'existe pas pour éviter l'erreur complète
    df = pd.DataFrame({'texte': [], 'sentiment': []})

# S'assurer qu'il y a des données et une colonne 'sentiment'
if not df.empty and 'sentiment' in df.columns:
    # 1. Split initial: 80% train, 20% test
    # 'stratify' est supprimé car une classe a trop peu de membres pour la stratification.
    train_df, test_df = train_test_split(
        df,
        test_size=0.2,
        random_state=42
    )

    # 2. Split du train: 80% train, 20% validation
    # 'stratify' est supprimé pour la même raison.
    train_df, val_df = train_test_split(
        train_df,
        test_size=0.2, # 20% de l'ensemble d'entraînement initial (soit 16% du total)
        random_state=42
    )

    print(f"Train: {len(train_df)} ({len(train_df)/len(df):.0%})")
    print(f"Validation: {len(val_df)} ({len(val_df)/len(df):.0%})")
    print(f"Test: {len(test_df)} ({len(test_df)/len(df):.0%})")
else:
    print("Impossible de diviser le dataset: 'df' est vide ou manque la colonne 'sentiment'.")
```

Étape 5 : Amélioration Continue

```python
# 05_analyse.py - Boucle d'Amélioration

# EXPLICATION : Boucle d'amélioration
# Le modèle est utilisé pour identifier les cas difficiles.
# L'humain corrige ces cas et les ajoute au dataset pour le prochain entraînement.

# NOTE : Supposons que 'model' est le modèle Naive Bayes entraîné de l'Étape 3.
# Supposons que 'df' est le dataset final chargé.

print("\n MODE EXPERT : ÉVALUATION DES PRÉDICTIONS")
nouvelles_phrases = [
    "Le chef est un artiste",
    "Je me suis senti arnaqué",
    "Assez bon dans l’ensemble"
]
corrections = []

for phrase in nouvelles_phrases:
    # EXPLICATION : Prédiction et Validation
    # Le modèle fait une prédiction sur le nouveau cas.
    prediction, _ = predire_sentiment_nb(phrase) # Utilise la fonction de l'Étape 3

    print(f"\nPhrase: {phrase}")
    print(f"Modèle: {prediction}")

    # Validation humaine interactive
    reponse = input("Correct? (o oui, n non, p partiellement): ")

    if reponse.lower() == 'n':
        # Le modèle s'est trompé : on collecte la correction.
        vrai_sentiment = input("Vrai sentiment? (positif/négatif/neutre): ")
        corrections.append({
            'texte': phrase,
            'prediction': prediction,
            'correction': vrai_sentiment, # Le nouveau label correct
            'source': 'correction_humaine',
            'qualite': 'haute'
        })

# Ajout des corrections au dataset
if corrections:
    # EXPLICATION : Enrichissement du Dataset
    # Les corrections sont converties en DataFrame et concaténées (ajoutées)
    # au dataset existant pour l'enrichir de nouveaux exemples de haute qualité.
    df_corrections = pd.DataFrame(corrections)
    df_ameliore = pd.concat([df, df_corrections], ignore_index=True)
    df_ameliore.to_csv('data/dataset_ameliore.csv', index=False)
    print(f"\n✅ {len(corrections)} corrections ajoutées et sauvegardées dans 'data/dataset_ameliore.csv'.")
else:
    print("\nAucune correction nécessaire. Le dataset est stable.")
```

9.2 Vérification de la Qualité

```python
def verifier_qualite_dataset(df):
    """Vérifie la qualité du dataset"""
    problèmes = []
    # 1. Vérifier les valeurs manquantes
    if df.isnull().any().any():
        problèmes.append("Valeurs manquantes détectées")
    # 2. Vérifier la distribution des classes
    distribution = df['sentiment'].value_counts(normalize=True)
    if distribution.min() < 0.1: # Moins de 10% dans une classe
        problèmes.append(f"Classe sous-représentée: {distribution.idxmin()}")
    # 3. Vérifier les doublons
    doublons = df.duplicated(subset=['texte']).sum()
    if doublons > 0:
        problèmes.append(f"{doublons} doublons détectés")
    # 4. Vérifier la longueur des textes
    if df['texte'].str.len().min() < 3:
        problèmes.append("Textes trop courts détectés")
    return problèmes
```
